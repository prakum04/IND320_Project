# Dashboard basics - part 1

#### Importing libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

##### Reading the CSV file

```python
df = pd.read_csv("../data/reservoirs.csv")
```

##### Rename headers to English and understandable

```python

#Creating a list with the english column names and setting as the column title
header = ["date_id"
,"area_type"
, "area_number"
, "iso_year"
,"iso_week"
, "fill_level"
,"capacity_TWh"
, "stored_energy_Twh"
, "next_publication_date"
, "fill_level_last_week"
, "change_in_fill_level"]

df.columns = header
```

##### Printing the data in a relevant way

```python
#Getting an overview of the data
df.head()
```

```python
#Checking column names, number of non-null values, data types and number of rows
df.info()
```

```python
#Gives a summary of numerical columns: mean, standard deviation, outliers, min and max. 
df.describe()
```

```python
#Missing values
df.duplicated().sum()
```

```python
df.nunique()
```

##### Plot each column separately 

```python
df["area_type"].value_counts().plot(kind="bar")
```

```python
# area_number: Identifies different geographical areas. A bar chart shows the number of observations per area.
df["area_number"].value_counts().plot(kind="bar")
```

```python
# area_type: Categorical variable. A bar chart shows the number of observations for each area type.
df["area_type"].value_counts().plot(kind="bar")
plt.title("Area Type")
plt.show()
```

```python

df["iso_year"].value_counts().sort_index().plot(kind="bar",  figsize=(12, 5))

plt.xlabel("Year")
plt.ylabel("Number of observations")
plt.title("Number of Observations per Year")
plt.show()
```

```python

df["iso_week"].value_counts().sort_index().plot(kind="bar",  figsize=(12, 5))

plt.xlabel("Week")
plt.ylabel("Number of observations")
plt.title("Number of Observations per Week")
plt.show()
```

```python
# fill_level: Continuous numerical variable. A histogram shows the distribution of reservoir fill levels.
df["fill_level"].plot(kind="hist")
plt.title("Fill Level")
plt.show()
```

```python
# capacity_TWh: Continuous numerical variable. A histogram shows the distribution of reservoir capacity.
df["capacity_TWh"].plot(kind="hist")
plt.title("Capacity (TWh)")
plt.show()
```

```python
# stored_energy_TWh: Continuous numerical variable. A histogram shows the distribution of stored energy.
df["stored_energy_Twh"].plot(kind="hist")
plt.title("Stored Energy (TWh)")
plt.show()
```

```python
# fill_level_last_week: Continuous numerical variable. A histogram shows the distribution of last week's fill levels.
df["fill_level_last_week"].plot(kind="hist")
plt.title("Fill Level Last Week")
plt.show()
```

```python
# change_in_fill_level: Continuous numerical variable. A histogram shows the distribution of changes in fill level.
df["change_in_fill_level"].plot(kind="hist")
plt.title("Change in Fill Level")
plt.show()
```

### Plotting Columns together

```python
#Looking at the correlation between strored energy Twh and capacity Twh. 

df.plot(

    x="capacity_TWh",
    y="stored_energy_Twh",
    kind="scatter"
)

plt.show()
```

```python
df.groupby("area_number")["fill_level"].mean().plot(kind="bar")

plt.xlabel("Area number")
plt.ylabel("Average fill level")
plt.title("Average Fill Level by Area")
plt.show()
```
