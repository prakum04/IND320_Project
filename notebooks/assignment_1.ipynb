#### All Columns together

The measurement variables have different original scales, so MinMaxScaler was used to scale them between 0 and 1. The data was grouped by date using the mean across reservoir areas. This allows the different measurement variables to be compared in the same plot.

The fill level and previous week's fill level follow very similar patterns, while the change in fill level fluctuates more. Capacity remains relatively constant over time.

```python
from sklearn.preprocessing import MinMaxScaler
import matplotlib.pyplot as plt

# Measurement columns to compare
columns_to_scale = [
    "fill_level",
    "capacity_TWh",
    "stored_energy_Twh",
    "fill_level_last_week",
    "change_in_fill_level"
]

# Convert date to datetime
df["date_id"] = pd.to_datetime(df["date_id"])

# Create a copy of the dataframe
df_scaled = df.copy()

# Scale measurement columns between 0 and 1
scaler = MinMaxScaler()

df_scaled[columns_to_scale] = scaler.fit_transform(
    df[columns_to_scale]
)

# Calculate the mean across reservoir areas for each date
df_plot = (
    df_scaled
    .groupby("date_id")[columns_to_scale]
    .mean()
)

# Plot all measurement columns together
#df_plot.plot(figsize=(14, 7))

df_plot.plot(
    figsize=(14, 7),
    style=["-", "-", "-", "--", "-"]
)

plt.title("Scaled Reservoir Data Over Time - Mean Across Areas")
plt.xlabel("Date")
plt.ylabel("Scaled Value (0–1)")
plt.legend(title="Variable")
plt.grid(alpha=0.3)
plt.tight_layout()

plt.show()
```
